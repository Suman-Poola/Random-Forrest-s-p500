```python
from Data_Frame import build_dataset

df, x_test, x_train, y_train, y_test = build_dataset()
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.tree import DecisionTreeRegressor
tree = DecisionTreeRegressor(max_depth = 5, random_state=72)
```

```python
x_train, x_test = x_test, x_train
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
features = [
    'close',
    'return',
    'Lag 1',
    'Lag 2',
    'Lag 3',
    'Lag 5',
    'volatility 5',
    'volatility 10',
    'volatility 20',
    'volatility 30',
    'SMA 10',
    'SMA 20',
    'SMA 30',
    'SMA 50',
    'volume',
    'Price change per volume',
    'id',
    'mk cap'
]

for feature in features:
    temp_x_train = x_train.drop(columns=[feature])
    temp_x_test = x_test.drop(columns=[feature])

    tree.fit(temp_x_train, y_train)

    predictions = tree.predict(temp_x_test)

    score = mean_squared_error(y_test, predictions)

    print(score, feature)
```
